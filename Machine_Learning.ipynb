#Square root of 10
The answer is between 3*3 = 9 and 4 * 4 = 16

```python
def mysqrt(nums, lo, hi, iterations):
  for i in range(iterations):
    guess = (lo + hi)/2
    if guess*guess > nums:
      hi = guess
    else:
      lo = guess
  return guess
result = mysqrt(10, 3, 4, 20)
print(f"sqrt(10)", result)
```

#Cuberoot
Cuberoot of 33

Answeer is between 3 and 4


```python
def mycuberoot(num, lo, hi, iterations):
  for i in range(iterations):
    guess = (lo + hi)/2
    if guess*guess*guess > num:
      hi = guess
    else:
      lo = guess
  return guess
result = mycuberoot(33, 3, 4, 20)
print(f"cuberoot(33))", result)
```

#Distance between Two points

Distance between two points = ((x2-x1)^2 + (y2-y1)^2) ^ 1/2

It is kind of hypotenuse where

base = x2-x1 (horizontal separation)
height = y2-y1 (vertical separation)

In Python

find_distance_2d(x1, y1, x2, y2) = sqrt((x2 - x1)**2 + (y2 - y1)**2)

```python
def find_distance_2pts(x1, y1, x2, y2):
  return sqrt((x2 - x1)**2 + (y2- y1)**2)

print(find_distance_2pts(0, 0, 3, 4))
print(find_distance_2pts(1, 2, 4, 6))
```

Distance between 3 points

```python
def find_distance_3pts(x1, y1, z1, x2, y2, z2):
  return sqrt((x2-x1)**2 + (y2-y1)**2 + (z2-z1)**2)
print(find_distance_3pts(0, 0, 0, 1, 1, 1))
print(find_distance_3pts(2, 3, 5, 2, 3, 5))
print(find_distance_3pts(1, 2, 3, 4, 6, 9))
```

#Predicting `y` from `x`

A straight line in 2D can be written as:

y = m * x + c {m is the slop (steepness) and c is the intercept(where line crosses the y-axis}

```python
def pred_y(m, c, x):
  return m * x + c
print(f"y : {pred_y(2, 3, 4)}")
```

```python
import matplotlib.pyplot as plt

def plot_line(m, c, x_range=(-10, 10), extra_points=None, title=None):
    n = 100
    xs = [x_range[0] + i * (x_range[1] - x_range[0]) / n for i in range(n + 1)]
    ys = [pred_y(m, c, x) for x in xs]

    plt.figure(figsize=(2, 3))
    plt.plot(xs, ys, 'b-', label=f"y = {m}x + {c}")
    if extra_points:
        px = [p[0] for p in extra_points]
        py = [p[1] for p in extra_points]
        plt.scatter(px, py, color='red', zorder=5, label='points')
    plt.axhline(0, color='gray', linewidth=0.5)
    plt.axvline(0, color='gray', linewidth=0.5)
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title(title or f"y = {m}x + {c}")
    plt.legend()
    plt.grid(True)
    plt.show()
```

#Rate of Change, Without Calculus

This is called the forward difference method:

f'(x) = ( f(x + h) - f(x) ) / h

where h is a small number

```python
def square(x):
    return x * x

def approx_derivative(func, x, h):
  return ( func(x + h) - func(x) ) / h

approx_derivative(square, 3, 0.0001)
```

Closer Point (Comparision between two Points)

```python
def closer_point(p, a, b):
  p_to_a = (p[0] - a[0])**2 + (p[1] + a[1])**2
  p_to_b = (p[0] - b[0])**2 + (p[1] + b[1])**2
  if p_to_a < p_to_b:
    return "A"
  else:
    return "B"

closer_point(p=(1, 2), a=(0, 0), b=(5, 5))
```
