```python
# Write code to calculate power 
def cal_pow(x,n):
    if n==1:
        return x
    else:
        return x*cal_pow(x, n-1)
cal_pow(2,0)    
```

```python
def power(x,n):
    if n==0:
        return 1
    else:
        return x*power(x, n-1)
power(2,0)    
```
