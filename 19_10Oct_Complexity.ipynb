```python
def find_min(arr):
    min = arr[0]
    for i in arr:
        if i < min:
            min = i
    return min
find_min([1,-5,3,1])   
#Complexity is eqyal to n     
```

```python
def print_n(n):
    total = 0
    for i in range(n):
        for j in range(n):
            print("hello")
            total += 1
print(print_n(2))            
```
