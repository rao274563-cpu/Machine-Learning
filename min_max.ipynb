Finding Minimum and Maximum

```python
def find_min(numbers):
    min = numbers[0]
    for num in numbers:
        if num < min:
            min = num
    return(min)  
print(find_min([5,8,2,6,1]))
      
```

```python
def find_max(numbers):
    max = numbers[0]
    for i in numbers:
        if i > max:
            max = i
    return(max)
print(find_max([5,6,4,1,8]))
        
    
```

```python
def find_min_max(numbers):
    min = numbers[0]
    max = numbers[0]
    for num in numbers:
        if num < min:
            min = num
        if num > max:
            max = num
    return(min), (max)  
print(find_min_max([5,8,1,2,3]))       
```

```python
#(a - min) / (max - min) used to max_min_normalization
```

```python
#min-max normalization
# samllet become 0 and biggest become 1 and everything in between proportionality
#            (a - min) / (max - min)
#[1, 3, 4] = [0, (3-1)/4-1) ,1] = [0, 2/3, (1]
```

```python
def min_max_normalization(value, numbers):
    mn, mx = find_min_max(numbers)
    return   (value - mn)/(mx - mn)
print(min_max_normalization(20, [10, 20, 30])) # (20 - 10)/(30 - 10) -> 10/20 = 0.5
```

```python
print(min_max_normalization(10, [10, 20, 30]))
```

```python
min_max_normalization(30, [10, 20, 30])
```

Compute Mean

```python
def compute_mean(numbers):
    if len(numbers) == 0:
        return 0
    return sum(numbers) / len(numbers)
print(f"Mean of [1,2,3,4] = ",compute_mean([1,2,3,4]))
```

```python
print(f"Mean of [10,20,30] = ",compute_mean([10,20,30]))
```

Compute Standard Deviation

```python
def std_deviation(numbers):
    mue = compute_mean(numbers)
    variance = compute_mean([(x - mue) ** 2 for x in numbers])
    return variance ** .5 
print(f"Standard Deviation = [10,20,30] = ",std_deviation([10,20,30]))
```
