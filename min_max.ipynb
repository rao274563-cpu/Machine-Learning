Finding Minimum and Maximum

```python
def find_min(numbers):
    min = numbers[0]
    for num in numbers:
        if num < min:
            min = num
    return(min)  
print(find_min([5,8,2,6,1]))
      
```

```python
def find_max(numbers):
    max = numbers[0]
    for i in numbers:
        if i > max:
            max = i
    return(max)
print(find_max([5,6,4,1,8]))
        
    
```

```python
def find_min_max(numbers):
    min = numbers[0]
    max = numbers[0]
    for num in numbers:
        if num < min:
            min = num
        if num > max:
            max = num
    return(min), (max)  
print(find_min_max([5,8,1,2,3]))       
```
