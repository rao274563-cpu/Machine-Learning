Finding Minimum and Maximum

```python
def find_min(numbers):
    min = numbers[0]
    for num in numbers:
        if num < min:
            min = num
    return(min)  
print(find_min([5,8,2,6,1]))
      
```

```python
def find_max(numbers):
    max = numbers[0]
    for i in numbers:
        if i > max:
            max = i
    return(max)
print(find_max([5,6,4,1,8]))
        
    
```

```python
def find_min_max(numbers):
    min = numbers[0]
    max = numbers[0]
    for num in numbers:
        if num < min:
            min = num
        if num > max:
            max = num
    return(min), (max)  
print(find_min_max([5,8,1,2,3]))       
```

Normalization : min become 0, max become 1 and middle become 0 <= mid <= 1

normalized(x) = (value - min)/(max-min) -> value will be the elements in the array -> same formula will used for each element normalization
normalize([10, 20, 30]) = (10 - 10)/(30-10),  (20 - 10)/(30-10),  (30 - 10)/(30-10) -> [0,  0.5,  1]

```python
def normalize(value, numbers):
    min, max = find_min_max(numbers)
    normalized = (value - min)/ (max - min)
    return normalized
print(normalize(10, [10, 20, 30]))    
```

Compute Mean : sum(elements)/no. of elements

```python
#print sum of number
def sum_num(numbers):
    sum = 0
    for num in numbers:
        sum += num
    return sum
print(sum_num([8,5,7]))    
```

```python
def compute_mean(numbers):
    sum = 0
    for num in numbers:
        sum += num
    mean = sum / len(numbers)
    return mean
print(compute_mean([8,4,3]))
```

```python
def compute_mean(numbers):
    sum = 0
    len = 0
    for num in numbers:
        sum += num
        len += 1
    mean = sum / len
    return mean
print(compute_mean([8,4,3]))
```

Compute Standard Deviation
mu = mean
variance = sum(xi - mu)**2 for i to n
std deviation = sqrt(variance/2)

```python
import math
def std_deviation(numbers):
    mu = compute_mean(numbers)
    variance = 0
    len = 0
    for i in numbers:
        len += 1         # Here the length is not count till last element, so we get wrong value
        variance += ((i - mu)**2) /  len 
    deviation = (variance)**.5 
    return deviation 
print(std_deviation([2, 4, 4, 4, 5, 5, 7, 9]))
```

Z-Score

Values with |z| > threshold are outliers.

z = abs((x - mu) / sigma) > threshold

```python
def find_z_score(value, numbers):
    mu = compute_mean(numbers)
    sigma = std_deviation(numbers)
    if sigma == 0:
        return 0
    return (value - mu)/sigma
print(find_z_score(10, [10, 20, 30]))
```
