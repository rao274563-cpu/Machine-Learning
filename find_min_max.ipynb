Finding Minimum and Maximum

```python
def find_min_max(numbers):
    current_min = numbers[0]
    current_max = numbers[0]
    
    for num in numbers:
        if num < current_min:
            current_min = num
        if num > current_max:
            current_max = num
    return(current_max, current_min)

print(find_min_max([5, 8, 2, 10, 3]))        
            
```
