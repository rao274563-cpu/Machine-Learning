```python
def myf(x):
    return x*x
```

```python
#Find gradient
x = 4
y = myf(x)

h = 0.0000000001
x1 = x + h
y1 = myf(x1)

change_in_x = y1 - y
change_in_y = x1 - x

grad = change_in_y/change_in_x

print(grad)
```

```python
h = .0000001
x = 4
y = myf(x)

#1. we increase x little bit
x1 = x + h
y1 = myf(x)

grad = (y1 - y)/(x1 - x)

#2. we decrwease x little bit.
x1 = x - h
y1 = myf(x)

grad = (y-y1)/(x-x1)
print("Grad2: ",grad)

```
