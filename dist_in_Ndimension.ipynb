Closure Point

```python
def sq_dist(A, B):
    x1, y1 = A
    x2, y2 = B
    dist = (x1 - y1)**2 + (x2 - x1)**2
    print(dist)
    return dist

def closer_point(P, A, B):
    pa = sq_dist(P, A)
    pb = sq_dist(P, B)
    print(pa,pb)
    if pa < pb:
        return "A"
    elif pa > pb:
        return "B"
    else: 
        return "Equal"   
print(closer_point((2,3),(1,2),(1,3)))    
```
