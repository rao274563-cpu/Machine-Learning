```python
def diff(f, x):
    eps = 1e-10
    d = x * 1e-10
    if d == 0:
        d = eps
    # return (f(x + d) - f(x))/d
    # return (f(x) - f(x-d))/d
    return (f(x+d) - f(x-d))/(2*d)
print(diff(lambda x: x**2, 1))
```

```python
import math
def f(x):
    return math.log(x)
print(diff(f, x=2)) # diff(log(x)) -> 1/2 * log(x) -> 1/2 *log2(2) -> 1/2 * 1 = 0.5
```

```python
def tax(income):
    if income > 10000:
        return .3 * (income - 10000)
    return 0
print(tax(12000))
```

Square distance

```python
def sqr_dist(x1, y1, x2, y2):
    A = (x1 - x2)**2
    B = (y1 - y2)**2
    return A + B

def sqr_dist(P, A, B):
    pa = sqr_dist(P, A)
    pb = sqr_dist(P, B)
    if pa > pb:
        return "B is closest to P"
    elif pa < pb:
        return "A is closest to P"
    else:
        return "Both are at equal distance from P"
```
