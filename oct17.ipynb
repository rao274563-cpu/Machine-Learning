```python
def multiply(x, y):
    if y == 0:
        return 0
    elif y == 1:
        return x 
    else:
        return x + multiply(x, y-1)
print(multiply(2,1))    
```

Function for Multiplication

```python
#Function fot multiplication
def mul(a,b):
    if b == 0:
        return 0
    elif b == 1:
        return a
    else:
        return a + mul(a, b-1)
print(mul(7,5))     
```

Function for Devision
0/1 = 0
1/0 = infinity

```python
def div(a,b):
    if a == 0 and b == 1:
        return 0
    elif a == 1 and b == 0:
        return "Infinity"
    else:
        return a/b
print(div(6,3))  
```

```python
def division(a,b):
    count = 0
    while a >= b:
        a = a-b
        count += 1
    return count,a
```

```python
print(division(4,2))
```

```python
division(18,3)
```

```python
division(32,6)
```
