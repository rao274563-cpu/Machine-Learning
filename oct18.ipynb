1. The rate of change of y/x is slope at any point.
2. WHen gradient (slope, dy/dx, differentiation, derivative) is zero, the curve could attain and minima or maxima
3. At minima, the gradient goes from -ive to 0 to +ive. So, the change in gradient is +ive. Gradient of gradient is +ive.
4. At maxima, the gradient goes from +ive to 0 to -ive. So, the change in gradient is -ive. Gradient of gradient is +ive. Second derivative is -ive.

```python
def myf(x):
    return x*x
```

```python
x = 4
y = myf(x) # y = 16

h = 0.0000000001
x1 = x + h
y1 = myf(x1) # y1 = x1*x1

change_in_y = y1 - y
change_in_x = x1 - x

ratio = change_in_y / change_in_x  #dy/dx
print(ratio)
```

```python
x = 5
y = myf(x)

h = 0.0000000001
x1 = x + h
y1 = myf(x1)

change_in_x = x1 - x
change_in_y = y1 - y

grad = change_in_y / change_in_x

print(grad)
```

```python
x = 4
y = myf(x)
h = 0.00000001
#1. we increase the x little bit
x1 = x + h
y1 = myf(x1)

grad = (y1 - y) / (x1 - x)
print("Grad1", grad)

#2. we decrease the x little bit

x1 = x - h
y1 = myf(x1)

grad = (y - y1) / (x - x1)
print("Grad2",grad)

#3. first we increase the x then decrease

x1 = x - h
y1 = myf(x1)

x2 = x + h
y2 = myf(x2)

#grad = (y2 - y1) / (x2 - x1)
grad = (y2 - y1) / (2*h)
print("Grad3", grad)
```

```python
def diff(myf, x):
    h=.00001
    y=myf(x)
    # print(y)
    x1=x+h
    y1=myf(x1)
    # print(y1)
    changeiny = y1-y
    changeinx = x1-x
    ratio = changeiny/changeinx
    return ratio
    # print(ratio)
```

```python
def diff(f, x):
    h = 1e-9
    return (f(x + h) - f(h))/h

# Find the value of x for which this function achieve minimum value
def myf(x):
    # Define a function x**4 - 10*x - 50
    return x**4 - 90*x - 100
```
